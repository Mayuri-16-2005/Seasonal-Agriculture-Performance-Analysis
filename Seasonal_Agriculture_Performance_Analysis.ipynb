## 1. Introduction

Agriculture plays an important role in the economy and is strongly affected by seasonal conditions. Factors such as rainfall, temperature, soil conditions, irrigation, fertilizer usage and crop selection can influence agricultural productivity and profitability.

This project analyzes agricultural data across three major seasons: Kharif, Rabi and Zaid. The analysis focuses on understanding seasonal differences in crop yield, production, resource usage, environmental conditions and economic performance.

Python-based data analysis and visualization techniques are used to identify meaningful patterns and generate data-driven insights that can support better agricultural planning and decision-making.


## 2. Problem Statement

Agricultural performance can vary significantly across different seasons because of changes in environmental conditions, water availability, crop selection and resource utilization.

Without proper analysis, it can be difficult to identify which seasons provide better productivity and profitability and how different agricultural factors are related to performance.

This project aims to analyze the agricultural dataset, compare performance across Kharif, Rabi and Zaid seasons, identify important patterns and provide data-driven recommendations for improving agricultural productivity and resource utilization.

Data cleaning was performed to improve the quality and reliability of the dataset before analysis.

## 11. Crop Analysis

Crop-level analysis was performed to compare the average yield of different crops.

The relationship between crop performance and season was also examined to identify which crops achieved higher or lower average yields across Kharif, Rabi and Zaid seasons.

This analysis helps understand differences in crop performance and supports better seasonal crop planning.

```python
crop_yield = (
    df.groupby("Crop")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
)

crop_yield
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=crop_yield.values,
    y=crop_yield.index
)

plt.title("Average Yield by Crop")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Crop")

plt.show()
```

```python
crop_season_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

crop_season_yield
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.show()
```

## 16. Key Findings

Based on the analysis performed on the agricultural dataset, the following key findings were identified:

1. Kharif recorded the highest average yield of 5.63 tonnes/ha, followed by Rabi at 5.04 tonnes/ha and Zaid at 4.64 tonnes/ha.

2. Kharif recorded the highest average production at 46.31 tonnes, while Zaid recorded the lowest average production at 38.89 tonnes.

3. Average rainfall was highest during Kharif at 849.20 mm and lowest during Zaid at 304.65 mm.

4. Zaid recorded the highest average temperature at 31.04°C, while Rabi recorded the lowest at 23.49°C.

5. Average fertilizer usage was relatively similar across the three seasons.

6. Zaid recorded the highest average water usage at 6,419.89 m³ despite having the lowest average yield.

7. Kharif recorded the highest average revenue at approximately ₹710,719.

8. Kharif recorded the highest average profit at approximately ₹178,914.65.

9. Zaid recorded an average loss of approximately ₹24,804.82.

10. Kharif recorded the highest average disease and pest risk at 54.47%.

11. Overall, Kharif showed the strongest overall agricultural and economic performance, while Zaid showed comparatively weaker performance.
